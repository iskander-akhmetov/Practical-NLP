# Chapter 1 Lab — Your First NLP Pipeline

Companion to *Modern Natural Language Processing Systems*, Chapter 1. Verifies your environment and builds a tiny end-to-end pipeline: tokenize → clean → count → plot, on a handful of public-domain sentences (no external downloads required).

```python
import re
from collections import Counter
import matplotlib.pyplot as plt

print("Environment OK")
```

## 1. A tiny corpus (public domain)

```python
corpus = [
    "NLP turns text into something a machine can act on.",
    "A machine that reads text still does not understand it the way a human does.",
    "Tokenization is the first step of almost every NLP pipeline.",
    "Every later chapter in this book adds one more layer of structure on top of tokens.",
]
corpus
```

## 2. Tokenize + clean

```python
def tokenize(text):
    return re.findall(r"[a-z']+", text.lower())

all_tokens = [tok for sent in corpus for tok in tokenize(sent)]
print(all_tokens[:15])
```

## 3. Count and plot word frequencies

```python
freqs = Counter(all_tokens)
most_common = freqs.most_common(10)
words, counts = zip(*most_common)

plt.figure(figsize=(8, 4))
plt.bar(words, counts)
plt.xticks(rotation=45, ha="right")
plt.title("Top word frequencies in the toy corpus")
plt.tight_layout()
plt.show()
```
